15.Write a Python program to separate even and odd numbers from the following tuple:
1.Create an empty list named even_numbers.
2.Create an empty list named odd_numbers.
3.Use a loop to check each number.
4.Add even numbers to even_numbers.
5.Add odd numbers to odd_numbers.
6.Print both lists.

```python
# Separate even and odd numbers from a tuple

numbers = (12, 7, 8, 15, 20, 33, 40, 51)

even_numbers = []
odd_numbers = []

for number in numbers:
    if number % 2 == 0:
        even_numbers.append(number)
    else:
        odd_numbers.append(number)

print("Even numbers:", even_numbers)
print("Odd numbers:", odd_numbers)
```
